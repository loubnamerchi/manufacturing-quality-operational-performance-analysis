# 07 - Visualization


```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
pd.set_option('display.max_columns', 50)

from src.data.load_data import load_csv
from src.analysis.business_analysis import defect_rate_by_bucket, maintenance_vs_quality
from src.visualization.charts import plot_bar, plot_correlation_heatmap
from src.utils.config import load_config, resolve_path

config = load_config()
df = load_csv(resolve_path(config['data']['processed_path']))
figures_dir = resolve_path(config['outputs']['figures_dir'])
figures_dir.mkdir(parents=True, exist_ok=True)

%matplotlib inline
```

## Chart 1: Overall defect status split



```python
counts = df['DefectFlagLabel'].value_counts()
fig = plot_bar(
    counts.index, counts.values,
    title='Overall Defect Status Split (n=3,240 records)',
    xlabel='Defect Status', ylabel='Number of Records',
    save_path=figures_dir / 'nb07_defect_status_split.png',
)
plt.show()
```

**Interpretation:** This chart makes the 84.0% / 16.0% imbalance immediately visible - the large majority of recorded production batches in this dataset are flagged as defective.

## Chart 2: Defect rate by maintenance-hours range


```python
maint = defect_rate_by_bucket(df, 'MaintenanceHours')
fig = plot_bar(
    maint['MaintenanceHours_range'], maint['defect_rate_pct'],
    title='Defect Rate (%) Rises with Maintenance Hours',
    xlabel='Maintenance Hours Range', ylabel='Defect Rate (%)',
    save_path=figures_dir / 'nb07_defect_rate_by_maintenance.png',
)
plt.show()
```

**Interpretation:** The upward trend is clear and consistent across every bucket - this is the most actionable pattern found in the dataset.

## Chart 3: Average quality score by maintenance-hours range


```python
mq = maintenance_vs_quality(df)
fig = plot_bar(
    mq['MaintenanceHours_range'], mq['avg_quality_score'],
    title='Average Quality Score Falls with Maintenance Hours',
    xlabel='Maintenance Hours Range', ylabel='Average Quality Score',
    save_path=figures_dir / 'nb07_quality_score_by_maintenance.png',
)
plt.show()
```

## Chart 4: Correlation heatmap (final version)



```python
fig = plot_correlation_heatmap(df.select_dtypes(include='number'), title='Manufacturing Dataset - Correlation Heatmap', save_path=figures_dir / 'nb07_correlation_heatmap.png')
plt.show()
```

## Summary

Four report-ready charts were generated and saved to `reports/figures/`:
1. `nb07_defect_status_split.png` - overall defect rate.
2. `nb07_defect_rate_by_maintenance.png` - the strongest identified defect driver.
3. `nb07_quality_score_by_maintenance.png` - the same relationship from the quality-score angle.
4. `nb07_correlation_heatmap.png` - the full correlation picture.
